3.1. Створіть зводну таблицю, яка покаже, скільки покупок (інвойсів) було зроблено в кожній з філій (`Branch`) за різними категоріями товарів. Запишіть таблицю в змінну `invoices_by_category` і виведіть змінну на екран.
Ця таблиця допоможе проаналізувати, в якій філії купують найбільше товарів кожної з категорій.

```python
invoice_by_category = df.pivot_table(index = 'Branch', columns= 'Product line', values= 'Invoice ID', aggfunc = 'count')
invoice_by_category.head()
```

В якому філіалі магазину Branch найбільше купують товарів Health & Beauty, і хто(Gender)?

```python
invoice_by_gender = df.pivot_table(index = ['Branch', 'Gender'], columns= 'Product line', values= 'Invoice ID', aggfunc = 'count')
invoice_by_gender['Health and beauty'].agg(['max', 'idxmax'])
```

```python
invoice_by_gender.loc['B', 'Health and beauty']
```

Цікаво, що в філіалі B здовров'ям та красою цікавляться більше чоловікі, можна розширити лінійки доглядової косметики, та різні добавки, протеїни

Перевіримо, а чи ці покупки прибуткові. Можливо чоловіки купують багато дешевих товарів, а жінки менше, але дорогих.

```python
df[df['Product line'] == 'Health and beauty'].groupby(by = ['Gender','Branch','Product line'])['Total'].sum().round(2)
```

Все ж таки в гілці B чоловікі абсолютні лідери і по кількості і по сумі чеку, можемо винагородити їх розширенням асортименту або ж подивитись найпопулярніші товари і влаштувати якусь акцію типу 2+1.

Який найпопулярніший спосіб оплати по філіям та містам, в залежності від статевої приналежності.

```python
df.groupby(by=['Branch', 'City', 'Gender'])['Payment'].agg(lambda x: x.value_counts().idxmax())
```

Бачимо що в двох філіях чоловікі викорисовують Ewallet, а жінки картки і cash. цікаво що філія C має найпопклярніший спосіб оплати готівковий, що цікаово для 21 сторіччя і треба перевірити чи не мають проблем там інші способи оплати.
