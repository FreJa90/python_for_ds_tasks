**Функція Seaborn: `sns.boxplot()`**

- Яка візуальна різниця між Pandas та Seaborn бокс-плотами?

Назви на осі x розтащовані по різному. В Pandas алфавітний порядок, в Seaborn від найкращої погоди до найгіршої. Також в Pandas треба рибирати автоматичний title. Загалом Seaborn яскравіший за змовчуванням, коли в Pandas треба буде додатково додавати стилі для більшої привабливості.

```python
weather_labels = {
    1: 'Ясно',
    2: 'Туман',
    3: 'Легкий дощ',
    4: 'Сильний дощ'
}

df['weather_label'] = df['weather'].map(weather_labels)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))


df.boxplot(column='count', by='weather_label', ax=axes[0])
axes[0].set_title('Оренди за погодою (Pandas)')
axes[0].set_xlabel('Погодні умови')
axes[0].set_ylabel('Кількість оренд')
plt.suptitle('')


sns.boxplot(data=df, x='weather_label', y='count', ax=axes[1])
axes[1].set_title('Оренди за погодою (Seaborn)')
axes[1].set_xlabel('Погодні умови')
axes[1].set_ylabel('Кількість оренд')

plt.tight_layout()
plt.show()
```
